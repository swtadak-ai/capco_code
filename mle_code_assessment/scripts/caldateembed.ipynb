```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
```

```python
def cal_cosin_Simi(df_crm_sub_1,df_cal_sub2):
    results = []
    all_text = (
        df_crm_sub_1["combinedcol_crm1"].tolist() +
        df_cal_sub2["combine_col"].tolist()
    )
    vectorizer = TfidfVectorizer()
    tfidf_matrix = vectorizer.fit_transform(all_text)
    crm_vectors = tfidf_matrix[:len(df_crm_sub_1)]
    cal_vectors = tfidf_matrix[len(df_crm_sub_1):]

    for i in range(len(df_crm_sub_1)):

        scores = cosine_similarity(
            crm_vectors[i],
            cal_vectors
        )

        best_idx = scores.argmax()

        best_score = scores[0][best_idx]
        if best_score > 0.70:
            results.append({
                "crm_id": df_crm_sub_1.iloc[i]["crm_id"],
                "matched_event": df_cal_sub2.iloc[best_idx]["cal_id"],
                "score": best_score
            })
    return results
```

```python
import re
import string
```

```python
def clean_text(text):
    text = str(text)
    text = re.sub(r'\b\d{1,4}[-/]\d{1,4}[-/]\d{1,4}\b',' ',text)
    text.translate(str.maketrans('','',string.punctuation))
    text=text.lower()
    text=re.sub(r'\s+',' ',text).strip()
    return text
```

```python
def cal_rapid_Fuzz(df_crm1,df_cal2):
    result= [] 
    crmlist = df_crm1['combinedcol_crm1'].tolist()
    callist = df_cal2['combine_col'].tolist()
    for i in range(0,len(crmlist)):
        for j in range(0,len(callist)):
            score = fuzz.ratio(
                crmlist[i].lower(),
                callist[j].lower()
            )
            # print('score is {}'.format(score))
            if score > 70.00:
                print('score is {}'.format(score))
                result.append({
                    'crm_id': df_crm1.iloc[i]['crm_id'],
                    'cal_id':df_cal2.iloc[j]['cal_id'],
                    'score':score})
                
    return result
```

```python
from sentence_transformers import SentenceTransformer
from sklearn.metrics.pairwise import cosine_similarity
```

```python
def sentence_embed(df_crm1,df_cal1):
    model = SentenceTransformer('all-MiniLM-L6-v2')
    crm_embeddings = model.encode(
    df_crm1["combinedcol_crm1"].tolist()
    )
    print('crm embeddingsi ',crm_embeddings)

    cal_embeddings = model.encode(
        df_cal1["combine_col"].tolist()
    )
    print('cal_embedding is ',cal_embeddings)
    results = []

    for i in range(len(df_crm1)):
        # for j in range(len(df_cal1)):
            scores = cosine_similarity(
                [crm_embeddings[i]],
                cal_embeddings
            )

            best_idx = scores.argmax()
            print('best_idx',best_idx)
            best_score = scores[0][best_idx]
            print('best score',best_score)
            if best_score > 0.80:
                results.append({
                    "crm_id": df_crm1.iloc[i]["crm_id"],
                    "matched_event": df_cal1.iloc[best_idx]["cal_id"],
                    "score": best_score,
                    "match":'true'
                })
            else:
                  results.append({
                    "crm_id": df_crm1.iloc[i]["crm_id"],
                    "matched_event": df_cal1.iloc[best_idx]["cal_id"],
                    "score": best_score,
                    "match":'false'
                })
                 
    return results
```
